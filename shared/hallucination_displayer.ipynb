```python

import csv
import pandas as pd
from IPython.display import HTML, display

p = 'sus_cases_b1_clip_hq.csv'
rows = []
with open(p, newline='', encoding='utf-8') as f:
    r = csv.DictReader(f)
    for row in r:
        rows.append(row)

df = pd.DataFrame(rows)

def to_bool(x):
    if isinstance(x, bool):
        return x
    if x is None:
        return False
    s = str(x).strip().lower()
    return s in {'true','1','t','yes','y'}

df['sus_any_qwen'] = df['sus_any_qwen'].map(to_bool)
df['sus_any_mplug'] = df['sus_any_mplug'].map(to_bool)

refs_cols = [c for c in df.columns if c.startswith('refs_pl')]
refs_col = refs_cols[0] if refs_cols else None

df = df[df['sus_any_qwen'] & df['sus_any_mplug']].copy()

cols = []
if 'img_path_q' in df.columns:
    cols.append('img_path_q')
if 'pred_pl_qwen' in df.columns:
    cols.append('pred_pl_qwen')
if 'pred_pl_mplug' in df.columns:
    cols.append('pred_pl_mplug')
if refs_col:
    cols.append(refs_col)

df = df[cols].copy()

if 'img_path_q' in df.columns:
    df.insert(0,'obraz', df['img_path_q'].apply(lambda p: f"<img src='{p}' width='1500'>"))
    df = df.drop(columns=['img_path_q'])

if refs_col and refs_col != 'refs_pl' and refs_col in df.columns:
    df = df.rename(columns={refs_col:'refs_pl'})

print(len(df))
display(HTML(df.to_html(escape=False, index=False)))
```
